```python
# ==========================================
# 2. Membaca File CSV Hasil Scraping
# ==========================================
print("\n📖 Membaca data dari file CSV yang telah disimpan...")
df_all_reviews = pd.read_csv(drive_path_all)
```

```python
# ==========================================
# 3. Filter & Export Ulasan Rating 1-2 Bintang
# ==========================================
# Filter ulasan dengan rating 1 dan 2 bintang dari data CSV yang dibaca
df_low_rating = df_all_reviews[df_all_reviews['score'].isin([1, 2])].copy()

# Reset index
df_low_rating = df_low_rating.reset_index(drop=True)

# Simpan file terpisah untuk rating 1-2 bintang
df_low_rating.to_csv(drive_path_low, index=False, encoding='utf-8')
print(f"✅ File ulasan rating 1-2 bintang disimpan di: {drive_path_low}")
```

```python
# ==========================================
# 4. Ringkasan Data & Total Ulasan
# ==========================================
total_all = len(df_all_reviews)
total_low = len(df_low_rating)
total_high = total_all - total_low
pct_low = (total_low / total_all * 100) if total_all > 0 else 0

print("\n" + "=" * 60)
print("📊 RINGKASAN DATA ULASAN (DIBACA DARI CSV)")
print("=" * 60)
print(f"Total semua ulasan       : {total_all:,}")
print(f"Ulasan rating 1-2 bintang: {total_low:,}")
print(f"Ulasan rating 3-5 bintang: {total_high:,}")
print(f"Persentase ulasan rendah : {pct_low:.2f}%")
print("=" * 60)

# Tampilkan distribusi rating dari data CSV
print("\n📈 Distribusi Rating Lengkap:")
print(df_all_reviews['score'].value_counts().sort_index())
```
